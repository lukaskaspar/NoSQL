```python

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# načtení dat
business = pd.read_json("yelp_business_v1.json", lines=True)
review = pd.read_json("yelp_review_v1.json", lines=True)
user = pd.read_json("yelp_user_v1.json", lines=True)

print("BUSINESS:", business.shape)
print("REVIEW:", review.shape)
print("USER:", user.shape)

display(business.head())
display(review.head())
display(user.head())
```

```python
# chybějící hodnoty
def missing_table(df):
    return pd.DataFrame({
        "missing_count": df.isna().sum(),
        "missing_pct": (df.isna().mean() * 100).round(2)
    }).sort_values("missing_count", ascending=False)

display(missing_table(business))
display(missing_table(review))
display(missing_table(user))
```

```python
# základní statistiky
display(business.describe(include=[np.number]))
display(review.describe(include=[np.number]))
display(user.describe(include=[np.number]))
```

```python
# business - státy
state_counts = business["state"].value_counts().head(15)
display(state_counts)

plt.figure(figsize=(10,5))
state_counts.plot(kind="bar")
plt.title("Top 15 států podle počtu podniků")
plt.xlabel("Stát")
plt.ylabel("Počet podniků")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
rating_counts = business["stars"].value_counts().sort_index()

plt.figure(figsize=(10,6))
rating_counts.plot(kind="bar", width=0.8)

plt.title("Rozdělení hodnocení podniků")
plt.xlabel("Hvězdičky")
plt.ylabel("Počet podniků")
plt.xticks(rotation=0)
plt.grid(axis="y", alpha=0.3)

plt.tight_layout()
plt.show()
```

```python
# business - open / closed
open_counts = business["is_open"].value_counts().sort_index()
display(open_counts)

plt.figure(figsize=(6,4))
open_counts.plot(kind="bar")
plt.title("Otevřené vs zavřené podniky")
plt.xlabel("is_open")
plt.ylabel("Počet")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

```python
# business - města
city_counts = business["city"].value_counts().head(15)
display(city_counts)

plt.figure(figsize=(10,5))
city_counts.plot(kind="bar")
plt.title("Top 15 měst podle počtu podniků")
plt.xlabel("Město")
plt.ylabel("Počet podniků")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
# business - kategorie
categories_series = business["categories"].dropna().str.split(", ")
all_categories = categories_series.explode()
top_categories = all_categories.value_counts().head(20)

display(top_categories)

plt.figure(figsize=(10,6))
top_categories.sort_values().plot(kind="barh")
plt.title("Top 20 kategorií podniků")
plt.xlabel("Počet podniků")
plt.ylabel("Kategorie")
plt.tight_layout()
plt.show()
```

```python
# review - rozdělení hvězdiček
review_stars = review["stars"].value_counts().sort_index()
display(review_stars)

plt.figure(figsize=(8,5))
review_stars.plot(kind="bar")
plt.title("Rozdělení hvězdiček v recenzích")
plt.xlabel("Hvězdičky")
plt.ylabel("Počet recenzí")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

```python
# review - délka textu
review["text_length"] = review["text"].fillna("").str.len()
display(review["text_length"].describe())

plt.figure(figsize=(8,5))
review["text_length"].plot(kind="hist", bins=30)
plt.title("Délka textu recenzí")
plt.xlabel("Počet znaků")
plt.ylabel("Počet recenzí")
plt.tight_layout()
plt.show()
```

```python
# review - useful/funny/cool
reaction_cols = ["useful", "funny", "cool"]
display(review[reaction_cols].describe())

plt.figure(figsize=(8,5))
review[reaction_cols].mean().plot(kind="bar")
plt.title("Průměrné reakce na recenze")
plt.xlabel("Typ reakce")
plt.ylabel("Průměr")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

```python
# review - roky
review["date"] = pd.to_datetime(review["date"], errors="coerce")
review["year"] = review["date"].dt.year

year_counts = review["year"].value_counts().sort_index()
display(year_counts)

plt.figure(figsize=(10,5))
year_counts.plot(kind="line", marker="o")
plt.title("Počet recenzí podle roku")
plt.xlabel("Rok")
plt.ylabel("Počet recenzí")
plt.tight_layout()
plt.show()
```

```python
# user - review_count
display(user["review_count"].describe())

plt.figure(figsize=(8,5))
user["review_count"].dropna().plot(kind="hist", bins=30)
plt.title("Počet recenzí na uživatele")
plt.xlabel("review_count")
plt.ylabel("Počet uživatelů")
plt.tight_layout()
plt.show()
```

```python
# user - yelping_since
user["yelping_since"] = pd.to_datetime(user["yelping_since"], errors="coerce")
user["yelping_year"] = user["yelping_since"].dt.year

yelping_counts = user["yelping_year"].value_counts().sort_index()
display(yelping_counts)

plt.figure(figsize=(10,5))
yelping_counts.plot(kind="line", marker="o")
plt.title("Počet uživatelů podle roku registrace")
plt.xlabel("Rok")
plt.ylabel("Počet uživatelů")
plt.tight_layout()
plt.show()
```

```python
# user - useful/funny/cool
user_reactions = ["useful", "funny", "cool"]
display(user[user_reactions].describe())

plt.figure(figsize=(8,5))
user[user_reactions].mean().plot(kind="bar")
plt.title("Průměrné reakce u uživatelů")
plt.xlabel("Typ reakce")
plt.ylabel("Průměr")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

```python
# propojení datasetů
business_with_reviews = review["business_id"].nunique()
all_businesses = business["business_id"].nunique()

print("Počet unikátních business_id v review:", business_with_reviews)
print("Počet unikátních business_id v business:", all_businesses)
print("Podíl businessů s recenzí:", round(business_with_reviews / all_businesses * 100, 2), "%")
```

```python
users_in_reviews = review["user_id"].nunique()
users_total = user["user_id"].nunique()

print("Počet unikátních user_id v review:", users_in_reviews)
print("Počet unikátních user_id v user:", users_total)
print("Podíl review userů v user datasetu:", round(users_in_reviews / users_total * 100, 2), "%")
```

```python
# business stars vs avg review stars
review_avg = review.groupby("business_id")["stars"].mean().reset_index(name="avg_review_stars")
business_small = business[["business_id", "name", "stars"]].rename(columns={"stars": "business_stars"})

merged = review_avg.merge(business_small, on="business_id", how="inner")
display(merged.head())
display(merged[["avg_review_stars", "business_stars"]].describe())

plt.figure(figsize=(6,6))
plt.scatter(merged["business_stars"], merged["avg_review_stars"])
plt.title("Business stars vs průměr recenzí")
plt.xlabel("business.stars")
plt.ylabel("avg review stars")
plt.tight_layout()
plt.show()
```
